```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/student_data.csv")
print("Dataset shape:", df.shape)
```

```python
print(df.head())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"],bins=20,kde=True,color="darkgreen")
plt.title("Disturbution of original grade(G3)")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="studytime",y="G3", data=df,palette="Set2")
plt.title("Final grade across study time travels")
plt.xlabel("Study Time(1=low, 4=high)")
plt.ylabel("Final grade")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="G1",y="G3",hue="sex",data=df,palette="husl")
plt.title("First period grade vs Final grade")
plt.xlabel("G1 (Period 1 grade)")
plt.ylabel("G3(Final grade)")
plt.show()
```

```python
sns.pairplot(df[["age","studytime","G1","G2","G3","sex"]],hue="sex",palette="coolwarm")
plt.suptitle("Pairwise relationship",y=1.02)
plt.show()
```

```python
numeric_cols=["age","studytime","failures","absences","G1","G2","G3"]
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_cols].corr(),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
